```python
import requests

def fetch_price(coin, currency):
    url = "https://api.coingecko.com/api/v3/simple/price"
    
    # The corrected parameter dictionary
    my_parameters = {
        "ids": coin,
        "vs_currencies": currency
    }
    
    response = requests.get(url, params=my_parameters, timeout=100)
    return response.json()
```

```python
print(fetch_price("bitcoin", "usd"))
```
